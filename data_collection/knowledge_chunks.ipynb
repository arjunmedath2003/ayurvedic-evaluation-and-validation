```python
import json
import pandas as pd
import glob, os
import ast
from collections import defaultdict
```

```python
chunks = []
with open("ayurveda.json") as f:
    data = json.load(f)

for formulation, info in data.items():
    for plant in info.get("ingredients", []):
        chunks.append({
            "text": f"Formulation {formulation} contains the plant {plant}.",
            "metadata": {
                "type": "formulation_ingredient",
                "formulation": formulation,
                "plant": plant,
                "source": "ayurveda.json"
            }
        })
```

```python
len(chunks)
```

```python
phyto = []

for file in glob.glob("phytochemicals/*.csv"):
    plant = os.path.basename(file).replace(".csv", "").replace("_", " ")
    df = pd.read_csv(file)
    df = df.drop_duplicates()
    for i, row in df.iterrows():
        phyto.append({
            "text": f"Plant {plant} contains the phytochemical {row['Phytos']}.",
            "metadata": {
                "type": "plant_phytochemical",
                "plant": plant,
                "phytochemical": row['Phytos'],
                "source": file,
                "row": i
            }
        })
```

```python
phyto[:10],len(phyto)
```

```python
chunks.extend(phyto)
```

```python
len(chunks)
```

```python
protein=[]

for file in glob.glob("proteins2/*.csv"):
    phyto = os.path.basename(file).replace(".csv", "").replace("_", " ")
    df = pd.read_csv(file)
    df = df.drop_duplicates()
    for i, row in df.iterrows():
        protein.append({
            "text": f"Phytochemical {phyto} targets the human protein {row['Protein']}.",
            "metadata": {
                "type": "phytochemical_target",
                "phytochemical": phyto,
                "protein": row['Protein'],
                "source": file,
                "row": i
            }
        })
```

```python
protein[:10],len(protein)
```

```python
chunks.extend(protein)
```

```python
len(chunks)
```

```python
import ast

pathway=[]
for file in glob.glob("pathway/*.csv"):
    formulation = os.path.basename(file).replace(".csv", "")
    df = pd.read_csv(file)
    df = df.drop_duplicates()
    for i, row in df.iterrows():
        for j in ast.literal_eval((row['clusters'])):
            pathway.append({
            "text": f"Protein {j} participates in KEGG pathway ({row['Pathway name']}).",
            "metadata": {
                "type": "protein_pathway",
                "p-value": row['p-value'],
                "formulation": formulation,
                "source": file,
                "row": i
            }
        })
```

```python
pathway[:10],len(pathway)
```

```python
chunks.extend(pathway)
len(chunks)
```

```python
with open("cluster.json") as f:
    clusters = json.load(f)

clus=[]

for formulation, proteins in clusters.items():
    for i in proteins:
        clus.append({
            "text": f"In formulation {formulation}, proteins {', '.join(i)} form a highly connected interaction cluster.",
            "metadata": {
                "type": "ppi_cluster",
                "formulation": formulation,
                "source": "cluster.json"
            }
        })
```

```python
clus[:10],len(clus)
```

```python
chunks.extend(clus)
len(chunks)
```

```python
with open("knowledge_chunks.json","w") as f:
    json.dump(chunks,f,indent=4)
```

```python
pathway2=[]
for file in glob.glob("pathway/*.csv"):
    formulation = os.path.basename(file).replace(".csv", "")
    df = pd.read_csv(file)
    df = df.drop_duplicates()
    for i, row in df.iterrows():
        pathway2.append({
            "text": f"{formulation} causes KEGG pathway {row['Pathway name']} with a p-value of {row['p-value']}.",
            "metadata": {
                "type": "formulation_pathway",
                "p-value": row['p-value'],
                "formulation": formulation,
                "source": file,
                "row": i
            }
        })
```

```python
pathway2[:10],len(pathway2)
```

```python
res=defaultdict(list)

for file in glob.glob("pathway/*.csv"):
    formulation = os.path.basename(file).replace(".csv", "")
    df = pd.read_csv(file)
    df = df.drop_duplicates()
    for i,row in df.iterrows():
        cluster = sorted(ast.literal_eval((row['clusters'])))
        if cluster not in res[row['Pathway name']]:
            res[row['Pathway name']].append(cluster)
```

```python
len(res)
```

```python
kegg_cluster = []

for kg,clusters in res.items():
    for j in clusters:
        kegg_cluster.append({
            "text": f"Cluster containing {','.join(j)} causes {kg} pathway",
            "metadata": {
                "type": "kegg pathways and clusters",
                "keggs pathway": kg,
                "cluster":",".join(j),
            }
        })
```

```python
len(kegg_cluster)
```

```python
with open("knowledge_chunks.json","r") as f:
    data = json.load(f)

data.extend(pathway2)
data.extend(kegg_cluster)

with open("knowledge_chunks2.json","w") as f:
    json.dump(data,f,indent=4)
```
